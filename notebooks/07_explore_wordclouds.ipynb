# Wordclouds

```python
import pandas as pd

from lana_nlp.dashboard.visualizations.comparisons import create_wordcloud
```

```python
df = pd.read_csv("../data/processed/song_level_stats.csv")
df.head()
```

```python
album_1 = "Honeymoon"
album_2 = "Ultraviolence"

album_1_lyrics = df[
    df["album"] == album_1
]["nlp_cleaned_lyrics"]

album_1_text = " ".join(
    " ".join(lyrics)
    for lyrics in album_1_lyrics
)
```

```python
wordcloud = create_wordcloud(album_1_text)
wordcloud
```
